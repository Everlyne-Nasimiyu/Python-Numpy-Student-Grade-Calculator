calculates the grades of students using Numpy arrays.

```python
import numpy as np

while True:
    # Ask the user to enter the number of students and the number of subjects
    while True:
        try:
            num_students = int(input("Enter the number of students: "))
            if num_students <= 0:
                print("The number of students must be a positive value (greater than 0). Please try again.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter an integer.")

    while True:
        try:
            num_subjects = int(input("Enter the number of subjects: "))
            if num_subjects <= 0:
                print("The number of subjects must be a positive value (greater than 0). Please try again.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter an integer.")

    # Create a numpy array to store the marks of each student in each subject, the array should have the shape(number of students, number of subjects)
    marks = np.zeros((num_students, num_subjects), dtype=float)

    #Ask the user to enter the marks of each student in each subject
    print("\n")
    print("Enter the marks of each student in each subject (out of 100):")
    for i in range(num_students):
        for j in range(num_subjects):
            while True:
                try:
                    mark = float(input(f"Enter marks for Student {i+1}, Subject {j+1}: "))
                    if 0 <= mark <= 100:
                        marks[i, j] = mark
                        break
                    else:
                        print("Marks must be between 0 and 100. Please try again.")
                except ValueError:
                    print("Invalid input. Please enter a number.")


    # Calculate the total marks for each student using the sum() function of Numpy
    total_marks = np.sum(marks, axis=1)

    # Calculate the percentage for each student
    percentage = (total_marks / (num_subjects * 100)) * 100

    # Calculate the grade for each student
    def calculate_grade(pct):
        if pct >= 90: return "A+"
        elif pct >= 80: return "A"
        elif pct >= 70: return "B+"
        elif pct >= 60: return "B"
        elif pct >= 50: return "C"
        else: return "F"

    grades = np.array([calculate_grade(p) for p in percentage])

    # Display the result for each student in a tabular format with columns for student name, total marks, percentage, and grade.
    print("\n-----Student Grade Report-----")
    print("| {:<15} | {:<12} | {:<12} | {:<8} |".format("Student Name", "Total Marks", "Percentage", "Grade"))
    print("|{:-<17}|{:-<14}|{:-<14}|{:-<10}|".format("", "", "", ""))
    for i in range(num_students):
        print("| {:<15} | {:<12.2f} | {:<11.2f}% | {:<8} |".format(
            f"Student {i+1}", total_marks[i], percentage[i], grades[i]
        ))
    print("------------------------------------------------------------------")

    # Ask user if they want to continue
    continue_choice = input("\nDo you want to calculate grades for more students? (yes/no): ").lower()
    if continue_choice != 'yes':
        print("Thank you for using the Grade Calculator. Goodbye!")
        break
```
